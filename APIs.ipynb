# WEB SCRAPING

```python
import requests
url="https://www.scrapethissite.com/pages/simple/"
resp=requests.get(url)
```

```python
if resp.status_code==200:
    with open("scrape_data/s_data.html","w",encoding="utf-8") as f:
        f.write(resp.text)
```

# BEAUTIFULSOUP

```python
pip install beautifulsoup4
```

```python
from bs4 import BeautifulSoup

with open("scrape_data/s_data.html","r") as f:
    html_content=f.read()
soup=BeautifulSoup(html_content,"lxml")
```

```python
soup.find("h1")
all_h3=soup.find_all("h3")

all_countries=[]

for h3 in all_h3:
    name=h3.get_text(strip=True)

    population=h3.find_next("div").select("span.country-population")[0].get_text(strip=True)
    print(h3.find_next("div").select_one("span.country-population").get_text(strip=True))

    # print(h3.attrs)
    
    all_countries.append([name,population])
    


```

```python
import pandas as pd
df=pd.DataFrame(all_countries,columns=["name","population"])
```

```python
df
```

```python
df.to_csv("cleaned_data/data.csv",index=False)
```

```python
import requests
from bs4 import BeautifulSoup
page_count=1
while True:
    url=f"https://quotes.toscrape.com/page/{page_count}/"
    resp=requests.get(url)
    soup=BeautifulSoup(resp.text,"lxml")
    quotes=soup.select("div.quotes")
    if not quotes:
        print("no valid pages anymore")
        break;
    with open(f"scrape_data/q_data{page_count}.html","w",encoding="utf-8") as f:
        f.write(resp.text)
    print(f"downloaded data from page{page_count}")
    page_count=page_count+1
    
```
