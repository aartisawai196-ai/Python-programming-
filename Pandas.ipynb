# PROPERTIES OF SERIES
## 1] Series are homogenous
## 2] We can perform vectorization
## 3] handle missing values with Nan
## 4] mutable values but immutable size

```python
# homogenous data
s=pd.Series([1,2,3,4,5])
print(s)


s1=pd.Series([1,2,3,4,5])
s2=pd.Series([10,20,30,40,50])
print(s)
print(s**2)
print(s1+s2)# vectorization


# mutable values
s1[0]=100
print(s1[0])
print(s1)
change=s1.drop(0)
print(s1)# immutable size
print(change)
```

# melt & pivote for reshaping
## converting wide data into long format
## melt(id_vars,value_var,var_name,value_name)
## pivot-> long to wide format
## pivot(index,columns,values)

```python
import pandas as pd
df=pd.DataFrame({
    "country":["USA","USA","India","India"],
    "year":[2020,2021,2020,2021],
    "sales":[100,120,90,110],
    "profit":[20,25,18,22],
    
})
```

```python
df
```

```python
melted_df=df.melt(
    id_vars=["country","year"],
    value_vars=["sales","profit"],
    var_name="matrics",
    value_name="value"
)
original=melted_df.pivot(
   index=["country","year"],#unique
  columns="matrics",
   values="value"
)
```

```python
melted_df
```

# merging and joining Data
## inner join
## Left join
## Right join
## Outer join

```python
# inner join =>merging two tables on the basis of parameter
df_customers=pd.DataFrame({
    "customer_id":[1,2,3,4],
    "name":["Adom","Bob","Charlie","Dave"]
})

df_orders=pd.DataFrame({
    "order_id":[101,102,103,104],
    "customer_id":[2,1,4,5],
    "amount":[250,120,300,180]
})

pd.merge(df_customers,df_orders, on="customer_id")#inner join
pd.merge(df_customers,df_orders, on="customer_id", how="left")#left join
pd.merge(df_customers,df_orders, on="customer_id", how="right")#right join
pd.merge(df_customers,df_orders, on="customer_id", how="outer")#outer join
```
